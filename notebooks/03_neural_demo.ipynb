# 03 — Neural networks demo (frame-level + sequence-level)

Trains the neural models from `prosody.models` on the same two tasks as the
classical demo. Requires `torch` (CPU is fine; MPS / CUDA auto-detected).

```
import pickle
from prosody.data import speaker_split
from prosody.train import train_frame_level_neural, train_sequence_level_neural
from prosody.evaluate import format_metrics

with open("artifacts/corpus.pkl", "rb") as f:
    corpus = pickle.load(f)
sp = speaker_split(corpus, val_speakers=["m2b"], test_speakers=["f3a"])
```

## Frame-level Conv1D (sequence-to-sequence)

```
nf = train_frame_level_neural(sp.train, sp.val, sp.test, epochs=8, batch_size=64)
for task, m in nf.test_metrics.items():
    print(f"  [{task:>10s}] {nf.model_name}: {format_metrics(m)}")
```

## Sequence-level CNN

```
ncnn = train_sequence_level_neural(sp.train, sp.val, sp.test, kind="cnn", epochs=8)
for task, m in ncnn.test_metrics.items():
    print(f"  [{task:>10s}] {ncnn.model_name}: {format_metrics(m)}")
```

## Sequence-level BiLSTM

```
nbil = train_sequence_level_neural(sp.train, sp.val, sp.test, kind="bilstm", epochs=8)
for task, m in nbil.test_metrics.items():
    print(f"  [{task:>10s}] {nbil.model_name}: {format_metrics(m)}")
```

## Note on checkpoint filenames

Each model writes to a unique path: `artifacts/frame_level_cnn.pt`,
`artifacts/sequence_level_cnn.pt`, `artifacts/sequence_level_bilstm.pt`.
The original notebook saved both CNN and RNN to `best_prosody_model.pth`,
which silently overwrote the CNN.
