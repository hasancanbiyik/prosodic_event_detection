# 01 — Preprocessing demo

Thin notebook that exercises the refactored `prosody.data` and `prosody.labels`
modules on the AutoRPT corpus. All real logic lives in `src/prosody/`.

Run from the repo root with the venv activated.

```
from prosody.data import load_corpus, summarize, speaker_split
from prosody.labels import ToBIParser
from dataclasses import asdict
import json
```

## Parse a single .ton file

```
parser = ToBIParser()
sample = next(iter((__import__("pathlib").Path("AutoRPT_Data/f1a/j")).glob("*.ton")))
events = parser.parse_file(sample)
print(f"{sample.name}: {len(events)} events")
print("Prominence:", sum(1 for e in events if e.event_type == "prominence"))
print("Boundary  :", sum(1 for e in events if e.event_type == "boundary"))
print("Other     :", sum(1 for e in events if e.event_type == "other"))
```

## Load and summarize the full corpus

```
corpus = load_corpus("AutoRPT_Data", progress=True)
stats = summarize(corpus)
print(json.dumps(asdict(stats), indent=2))
```

## Speaker-aware split (the fix)

```
sp = speaker_split(corpus, val_speakers=["m2b"], test_speakers=["f3a"])
for split_name, files in [("train", sp.train), ("val", sp.val), ("test", sp.test)]:
    speakers = sorted({c.speaker for c in files})
    print(f"{split_name:5s}: {len(files):3d} files, speakers={speakers}")
```
